# Introduction to practical near-term quantum computing

Ever since quantum computers were first proposed more than four decades ago, the question has been: How far could this new approach extend the limits of computation? Theoretical computer science gave us some exciting answers. Quantum computers could factor large integers dramatically faster than the best-known classical methods, speed up unstructured search, and simulate quantum systems more naturally. However, many of these ideas assumed a future large-scale, fault-tolerant quantum computer.

The technology hasn't yet reached all of those theoretical goals, but it is improving rapidly. Researchers can now use quantum processors to explore real scientific and industrial problems. In this course, we'll get hands-on experience with some of the most promising applications. We'll estimate low-energy properties of a complex molecular model, simulate the dynamics of a lattice gauge theory, and explore how a financial portfolio problem can be mapped to a quantum computer. We'll also look at newer areas where we've only begun to understand what quantum computers might offer, including quantum machine learning (QML) and quantum approaches to partial differential equations (PDEs).

Many of the examples in this course are small enough for a classical computer to solve. That's intentional. The goal is to give you tangible experience with the full process of using a quantum computer to address a real-world problem. Some examples represent problems for which quantum advantage might emerge at much larger scales. In other examples, we use a simplified method so that implementation details don't hide the central quantum idea.

By the end of the course, you should have an overview of the main quantum applications and an intuition for why quantum computers might be useful in particular areas. You'll encounter common techniques for representing problems on quantum computers, learn how quantum and classical methods work together, and begin to recognize which tools and tradeoffs matter for different kinds of problems.

In this first lesson, we'll cover some of the fundamental concepts and terminology that help you understand applications of quantum computing at a high level, and give you an overview of the main application areas that the scientific community is actively pursuing.

## What quantum computers might be useful for

Quantum computers are often framed as competitors to classical computers, but a more appropriate way to think about them is as specialized resources within larger classical workflows. The important question, then, is which parts of a computation are especially well suited to quantum hardware. The most intuitive answer comes from problems that are already quantum, such as the behavior of molecules, materials, and quantum fields. A quantum processor can represent and manipulate quantum states directly, while the resources required for a general classical description grow exponentially with the size of the system.

The case is more subtle for problems with classical inputs and outputs, but there are still good reasons to explore them. Optimization problems are widespread and economically important, so even a modest improvement in solution quality or runtime could be valuable. Quantum algorithms might also offer new ways to explore complicated solution spaces or exploit structure in particular classes of problems. Quantum machine learning has strong theoretical motivation for certain problems with classical inputs, including structured classification and the processing of massive datasets. However, whether these results can translate into practical advantage on real-world classical data remains an open question. Partial differential equations describe many important systems in science, engineering, and finance, and quantum algorithms might improve the scaling of certain high-dimensional simulations and large linear-algebra problems.

In all three of these areas, however, the cost of preparing classical inputs, running the quantum computation, and extracting a useful classical answer can outweigh the benefit of the quantum step. We'll keep these practical challenges in mind as we explore how quantum methods are applied in each area.

## Near-term quantum utility and advantage

This course isn't specifically about quantum advantage, but a discussion of quantum advantage is unavoidable. The fundamental motivation behind quantum applications is to use quantum resources to extend what classical computers can do alone: to produce more accurate results, reach solutions faster, or solve problems that are otherwise inaccessible. The following related terms help us talk clearly about the progress we see:

- **Quantum speedup:** As problem size increases, the time required by a quantum method scales more favorably than that of the best relevant classical method for the same task.

- **Quantum advantage:** A quantum-enabled computation outperforms the best relevant classical approaches on a well-defined task and meaningful metric. Rather than a single, permanent milestone, quantum advantage is an evolving frontier that shifts as both quantum and classical methods improve.

- **Quantum utility:** A quantum computation contributes to a useful scientific or technical result. It operates at a scale, depth, and accuracy beyond practical brute-force classical simulation, without necessarily outperforming the best classical method.

- **Near-term algorithm:** An algorithm designed for current quantum processors, using limited circuit depths and often relying on classical post-processing and error mitigation rather than full quantum error correction.

- **Fault-tolerant quantum computing:** In the future, quantum computing will use quantum error correction to protect logical qubits throughout a computation. When physical error rates are below the fault-tolerance threshold, additional error-correction resources can suppress logical errors to the level that long, reliable computations require. Fault tolerance makes much deeper algorithms possible, including large-scale implementations of Shor's algorithm.

- **Early fault tolerance:** Fault tolerance means that errors are suppressed, not eliminated entirely. In early fault-tolerant systems, residual logical errors must still be accounted for, and the depth of reliable circuits remains limited.

- **Heuristic algorithm:** A method designed to find a good solution in practice without guaranteeing the optimal answer or a particular runtime.

## A brief foundation in complexity theory

In some application areas, we have a strong theoretical reason to expect that a quantum computer might have an advantage. To understand where that confidence comes from, we need a little vocabulary from complexity theory.

Complexity theory studies how the resources needed to solve a problem grow as the problem gets bigger. It gives us a useful map of where a quantum speedup might be possible. A few terms are especially important for this course:

- **P:** Problems that a classical computer can solve efficiently, meaning that the required resources grow polynomially with the size of the problem.

- **NP:** Problems for which you can efficiently check a proposed solution, even if finding that solution might be very difficult.

- **NP-hard:** Problems that are at least as hard as every problem in NP. An NP-hard problem does not necessarily belong to NP. Many optimization problems are NP-hard.

- **NP-complete:** Problems that are both NP-hard and in NP. These are the hardest decision problems within NP.

- **BQP:** Problems that a quantum computer can solve efficiently with a bounded probability of error.

Some relationships among these categories are known. P is contained in both NP and BQP because a problem that can be solved efficiently on a classical computer can also be checked classically and solved on a quantum computer. Other relationships remain open. We don't know whether P equals NP or whether NP is contained in BQP. However, researchers generally believe that P is smaller than both NP and BQP, and that quantum computers cannot efficiently solve every NP-complete problem. In plain terms, a quantum computer is not a magic box that instantly solves every hard problem.

The strongest complexity-based arguments for quantum computing apply to particular kinds of problems. Shor's algorithm scales dramatically better than the best-known classical algorithms for factoring and discrete logarithms, although factoring has not been proven to be outside P. General quantum simulation also has a strong theoretical foundation. Some quantum-simulation problems are BQP-complete, meaning that efficiently solving them classically would amount to efficiently simulating any quantum computation. Such an efficient classical simulation is highly implausible, which is one reason that quantum Hamiltonian simulation is considered such a promising application area.

Some high-dimensional PDEs can be reduced to BQP-complete linear-system problems, giving us a strong reason to expect a quantum speedup when inputs and outputs can be handled efficiently. QML has similar examples: some tasks involving quantum data are BQP-complete, and some quantum models can generate samples that are difficult to reproduce classically.

The case is less straightforward for optimization. Quantum computers are not expected to solve arbitrary NP-hard problems efficiently, but they might still help with structured problems, particular instances, or specific steps in an optimization workflow. Grover's algorithm, for example, gives a quadratic speedup for unstructured search. That is a meaningful improvement, even though the square root of an astronomically large search space can still be astronomical.

## From complexity theory to practical evidence

Complexity theory can tell us where a quantum speedup might be possible, but it usually can't tell us how a method performs on a particular problem or a particular quantum computer. Many of the methods used on current hardware are heuristic. They are designed to find good answers in practice, but they do not guarantee the best possible answer or a particular runtime.

Heuristics are not unique to quantum computing, and calling a method heuristic does not mean that it is unreliable or unscientific. Classical optimization, machine learning, and computational chemistry all rely heavily on heuristics when an exact solution would be too expensive. A quantum heuristic might offer a new way to explore a solution space, prepare useful states, or identify promising candidates. Whether it works well is an empirical question: we have to run it and measure the result.

Because heuristics don't come with performance guarantees, we evaluate them using a clear success metric, strong classical baselines, and tests across different problem sizes. This helps us see where a method works well and how its performance changes as the problem scales.

## Application areas and course case studies

We organize the course around four broad application areas. They don't cover every possible use of a quantum computer, but they give us a useful way to compare where the strongest arguments lie and where the open questions are.

### Hamiltonian simulation

Hamiltonian simulation includes estimating energies, studying chemical and material properties, and modeling how quantum systems change over time. It reaches across chemistry, materials science, condensed-matter physics, and high-energy and nuclear physics. Our case studies examine a strongly correlated iron-sulfur model using sample-based quantum diagonalization and simulate hadron dynamics using a symmetry-aware lattice representation. These are two examples from a much larger field, which also includes reaction mechanisms, magnetic materials, and other many-body dynamics.

### Optimization

Optimization appears everywhere from finance and logistics to scheduling and energy systems. It is also an area in which advantage is difficult to establish, because classical solvers are extremely capable and quantum mappings can add costly overhead. Our case study uses portfolio selection to explore how an optimization problem is represented, run, and benchmarked without assuming that a quantum computer automatically finds the exact optimum.

### Quantum machine learning

Quantum machine learning asks whether quantum circuits or quantum data can improve a learning task. Important questions include data encoding, training cost, generalization, and comparison with strong classical models.

### Partial differential equations

Partial differential equations describe systems such as fluids, heat, and waves. Quantum algorithms might accelerate parts of some PDE workflows, but data preparation and extracting a useful classical answer can be major costs. The specific case study is still being developed.

In the lessons that follow, we'll ask the same basic questions in each area: What is the problem, why might quantum computing help, what are the limitations, and what does the evidence actually show? We'll start with Hamiltonian simulation, where the problem itself is quantum.

## Conclusion

We've covered a lot of ground, but the main idea is simple: quantum computers are specialized tools, and their value depends on using the right quantum method for the right problem or piece of a larger problem. Hamiltonian simulation has the clearest theoretical foundation, optimization has a mature set of heuristic methods and benchmarks, and QML and PDEs offer promising directions that are still developing.

In the lessons that follow, we'll move from this broad overview to hands-on examples. For each application, we'll ask what problem we're trying to solve, what role the quantum computer plays, how the result is interpreted, and what we can learn from it. The goal is to build your intuition for where quantum computing might help and what it takes to turn a promising idea into a real computation.
